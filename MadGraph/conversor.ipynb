# Conversor de archivos .root
En este notebook se realizaran los procesos de conversión de los archivos ```.root``` generados por el simulador de Delphes y se convertirán en archivos ```.root``` de tabla plana.

````python
import UpROOT.io as io
import UpROOT.config as conf
import UpROOT.delphes_root as dr
````

````python
collections = (
    'EFlowTrack',
    'Electron',
    'Event',
    'Jet',
    'Muon',
    'Particle',
    'Photon',
    'ScalarHT',
    'Track',
    'Weight'
)
````

## Archivos con modelo ```sm```

### Archivo 01 ($mmll, mmnl = 10GeV$)

````python
# mpmm_50k_10 (100k - concatenados)
ANALYSIS = ''
cfg = conf.CONFIGURATIONS[ANALYSIS]
DATASET = cfg['raw_dataset']
````

````python
io.load_dataframe(
    branches=('Muon/Muon.D0', 'Muon/Muon.DZ', 'Muon/Muon.ErrorD0', 'Muon/Muon.ErrorDZ'),
    dataset=DATASET,
).head()
# print(df.to_latex(index=True, bold_rows=True))
````

````python
output = dr.convert_dataset(
    dataset=DATASET,
    collections=collections,
    step_size='20 MB',
    object_registry=True,
    overwrite=True,
    create_dimuons=False
)
````
