# $B^{\pm} \to J/\psi K^{\pm}$ analysis example for LHCb collision ntuples.

This analysis example demonstrates how to work with LHCb collision
ntuples produced by the LHCb Ntupling Service. The example studies
$B^{\pm} \to J/\psi K^{\pm}$ decay and demonstrates plotting of the $B^{\pm}$ meson
candidate mass and $J/\psi$ candidate mass.

```python
import ROOT
from IPython.display import Image, display
ROOT.gROOT.SetBatch(True)

root_files = [
    "../../data/00334560_00000001_1.dvntuple.root",
    "../../data/00334560_00000002_1.dvntuple.root",
    "../../data/00334564_00000001_1.dvntuple.root",
    "../../data/00334565_00000001_1.dvntuple.root",
    "../../data/00334566_00000001_1.dvntuple.root",
]
rdf = ROOT.RDataFrame("Btree/DecayTree", root_files)
```

```python
hist_args_B = (
    "",
    "B candidate mass;#it{m}_{#mu^{+}#mu^{-}K^{#pm}};candidates",
    200,
    4600,
    6000,
)
hist_B = rdf.Histo1D(hist_args_B, "Bplus_M")
```

```python
hist_args_jpsi = (
    "",
    "J/#psi candidate mass;#it{m}_{#mu^{+}#mu^{-} [MeV/c^{2}]};candidates",
    200,
    2950,
    3250,
)
hist_jpsi = rdf.Histo1D(hist_args_jpsi, "J_psi_1S_M")
```

```python
canvas = ROOT.TCanvas("canvas", "", 900, 600)
hist_B.Draw()
canvas.SaveAs("figs/ej01_bmass.png")
hist_jpsi.Draw()
canvas.SaveAs("figs/ej01_jpsimass.png")
```

## Combinación de ajustes

### Ajuste gaussiano + polinomio lineal

```python
fit_gauspol1 = ROOT.TF1(
    "fit_gauspol1",
    "gaus(0) + pol1(3)",
    5150,
    5400
)
fit_gauspol1.SetParameters(
    10000, 5280, 20,    # función de gaus; amplitud, media, sigma
    100, -0.01          # función de pol1: intercepto, pendiente
)

hist_B.Draw()
fit_R_gauspol1 = hist_B.Fit(fit_gauspol1, "S", "", 5000, 5600)
canvas.SaveAs("figs/ej01_fit_gp1_bmass.png")
display(Image("figs/ej01_fit_gp1_bmass.png"))
```

Para la sintaxis `gaus(0) + pol1(3)` se tiene que `gaus(0)` usa parámetros `[0]`, `[1]`, `[2]` y `plo1(3)` usa parámetro `[3]`, `[4]`. Es decir:
* `[0]`: amplitud de la gaussiana.
* `[1]`: media de la gaussiana.
* `[2]`: sigma de la gaussiana.
* `[3]`: contraste del fondo.
* `[4]`: pendiente del fondo.

### Ajuste gaussiano + fondo exponencial

```python
fit_gausexp = ROOT.TF1(
    "fit_gausexp",
    "gaus(0) + expo(3)",
    5150,
    5400
)
fit_gausexp.SetParameters(
    10000, 5280, 20,    # función de gaus; amplitud, media, sigma
    100, -0.02          # función de expo: intercepto, pendiente
)

hist_B.Draw()
fit_R_gausexp = hist_B.Fit(fit_gausexp, "S", "", 5000, 5600)
canvas.SaveAs("figs/ej01_fit_gex_bmass.png")
display(Image("figs/ej01_fit_gex_bmass.png"))
```

Interpretación:
* `gaus(0)`: representa la señal del $B$.
* `expo(3)`: representa el fondo combinatorio.

### Ajuste con dos funciones gaussianas

```python
fit_gaus2 = ROOT.TF1(
    "fit_gaus2",
    "gaus(0) + gaus(3)",
    5150,
    5400
)
fit_gaus2.SetParameters(
    8000, 5280, 15,    # Primera gaussiana; amplitud, media, sigma
    3000, 5280, 35,    # Segunda gaussiana: amplitud, media, sigma
)

hist_B.Draw()
fit_R_gaus2 = hist_B.Fit(fit_gaus2, "S", "", 5000, 5600)
canvas.SaveAs("figs/ej01_fit_g2_bmass.png")
display(Image("figs/ej01_fit_g2_bmass.png"))
```

La gaussiana estrecha `gaus(3)` modela el núcleo del pico mientras que la gaussiana ancha `gaus(0)` de la resolución o cola.

### Doble gaussiana + fondo

```python
fit_gaus2expo = ROOT.TF1(
    "fit_gaus2expo",
    "gaus(0) + gaus(3) + expo(6)",
    5150,
    5400
)
fit_gaus2expo.SetParameters(
    8000, 5280, 15,    # Primera gaussiana; amplitud, media, sigma
    3000, 5280, 35,    # Segunda gaussiana: amplitud, media, sigma
    10, -0.001          # función de expo: intercepto, pendiente
)

hist_B.Draw()
fit_R_gaus2expo = hist_B.Fit(fit_gaus2expo, "S", "", 5000, 5600)
canvas.SaveAs("figs/ej01_fit_g2e_bmass.png")
display(Image("figs/ej01_fit_g2e_bmass.png"))
```

Los parámetros de los ajustes corresponden a:
* `[0]`, `[1]`, `[2]`: Primera gaussiana
* `[3]`, `[4]`, `[5]`: Segunda gaussiana
* `[9]`, `[7]`       : Fondo exponencial

```python
fit2_gaus2expo = ROOT.TF1(
    "fit2_gaus2expo",
    "gaus(0) + gaus(3) + expo(6)",
    5150,
    5400
)
fit2_gaus2expo.SetParameters(
    8000, 5280, 15,    # Primera gaussiana; amplitud, media, sigma
    3000, 5280, 35,    # Segunda gaussiana: amplitud, media, sigma
    10, -0.0001          # función de expo: intercepto, pendiente
)

hist_B.Draw()
fit2_R_gaus2expo = hist_B.Fit(fit2_gaus2expo, "S", "", 5000, 5600)
canvas.SaveAs("figs/ej01_fit2_g2e_bmass.png")
display(Image("figs/ej01_fit2_g2e_bmass.png"))
```

### Breit-Wigner
Se usa mucho para resonancias físicas con anchura natural.

```python
fit_bw = ROOT.TF1(
    "fit_bw",
    "[0]*TMath::BreitWigner(x, [1], [2])",  # [0]: amplitud, [1]: posición central, [2]: anchura
    5150,
    5400
)
fit_bw.SetParameters(
    10000, 5280, 20,
)

hist_B.Draw()
fit_R_bw = hist_B.Fit(fit_bw, "S", "", 5000, 5600)
canvas.SaveAs("figs/ej01_fit_bw_bmass.png")
display(Image("figs/ej01_fit_bw_bmass.png"))
```
