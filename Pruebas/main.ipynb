```python
import ROOT
ROOT.gROOT.SetBatch(True)
```

```python
file_path = "../data/00334560_00000001_1.dvntuple.root"

f = ROOT.TFile.Open(file_path)

if not f or f.IsZombie():
    raise RuntimeError("No se puede abrir el archivo ROOT")
tree = f.Get("Btree/DecayTree")

if not tree:
    raise RuntimeError("No se encontró el árbol Btree/Decay/Tree")
    
print("Número de eventos/candidatos:", tree.GetEntries())
print("\nEstructura completa del árbol:\n")
tree.Print()
```

```python
root_files =[
    "../data/00334560_00000001_1.dvntuple.root"
]
rdf = ROOT.RDataFrame("Btree/DecayTree", root_files)
columns = rdf.GetColumnNames()
print("Columnas disponibles\n")
for col in columns:
    print(col)
```
